**Evaluar multicolinealidad.**

````python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

def calcular_vif(df_independiente):
    # Crear un DataFrame para almacenar los resultados del VIF
    vif_data = pd.DataFrame()
    vif_data["Variable"] = df_independiente.columns
    vif_data["VIF"] = [variance_inflation_factor(df_independiente.values, i)
                       for i in range(df_independiente.shape[1])]
    return vif_data
````

````python
vif_resultado = calcular_vif(x)
print(vif_resultado)
````

Para evaluar la multicolinealidad vimos que hay tres tipos que son:

•  VIF = 1: No hay correlación entre la variable y el resto de predictores.

•  VIF entre 1 y 5: Correlación moderada, generalmente aceptable en la mayoría de los modelos.

•  VIF > 5 (o > 10): Alta correlación. La variable presenta multicolinealidad
Y con el resultado que nos salió fue vif y eso significa que no hay correlación entre la variable y el resto de predictores

## Modelo de regresión lineal múltiple.

````python
import statsmodels.api as sm

# Add a constant to x for the intercept term in the model.
# The 'x' in the current kernel state is the version without the constant (from cell 'j1wdAHBhnaPJ').
x_with_constant = sm.add_constant(x)

# Create and fit the linear regression model.
# This defines 'modelo_regresion'.
modelo_regresion = sm.OLS(y, x_with_constant).fit()

# Now that the model is defined, we can use it for prediction.
# We use 'x_with_constant' to ensure the prediction uses the same structure as the model fitting.
y_calculada = modelo_regresion.predict(x_with_constant)

print("Parámetros del modelo:")
print(modelo_regresion.params)
````

Para la parte de modelo de regresión aparecen las variables de peso que fue 1.237513 y de la parte de tipo de entrenamiento fue: 13.508878 y con la constante de 794.339145
