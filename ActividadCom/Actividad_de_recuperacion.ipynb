**VARIABLES DEL MODELO**

Como pirimer modelo vamos a utilizar una variable independiente que seria la edad y como variable dependiente es la frecuencia cardiaca

```python
Y = df["BS"]      # Variable dependiente
X = df["HeartRate"]    # Variable independiente
```

```python
df=df[(10<df['BS']) & (df['BS']<14)]
```

```python
# Importamos la librería matplotlib.pyplot y le damos el pseudónimo plt
import matplotlib.pyplot as plt

# Definir las variables X y Y a partir del DataFrame df
X = df['Age']
Y = df['HeartRate']

# Configuración general del gráfico
plt.figure(
    figsize=(6, 4),   # tamaño de la figura (ancho, alto) en pulgadas
    dpi=120           # resolución del gráfico
)

# Gráfico de dispersión
plt.scatter(
    X, Y,
    marker="o",       # forma: googlear "matplotlib.markers"
    color='#F527D3',     # color de los puntos
    edgecolor='black',    # borde de los puntos
    alpha=0.9,            # transparencia: 0 es invisible y 1 es color sólido
    s=50,                 # tamaño de los puntos
)

# Etiquetas de los ejes
# eje x
plt.xlabel(
    'Age', # etiqueta del eje x
    fontsize=12  # tamaño de fuente
)

plt.ylabel(
    'HeartRate', # etiqueta del eje y
    fontsize=12 # tamaño de fuente
)

# Fuente de los ticks
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# Márgenes
plt.gca().spines['right'].set_visible(False) # derecha
plt.gca().spines['top'].set_visible(False)   # superior

# Cuadrícula (opcional, pero didáctica)
plt.grid(
    visible=True,
    linestyle='--',
    linewidth=0.7,
    alpha=0.5,
    color="gray"
)
```

```python
import pandas as pd
import statsmodels.api as sm

# The dataframe 'df' is already loaded in a previous cell (tprkEPFaze4U).
# The variables X (Age) and Y (HeartRate) are already defined globally in a previous cell (qsiJ_TiY0e83).

# Add a constant to the independent variable
x_constante = sm.add_constant(X)

# Create and fit the OLS model
modelo = sm.OLS(Y, x_constante).fit()

# Print the model parameters
modelo.params
```

La ecuación de la recta es:

$$
\hat{y}=73.304304+0.037810X
$$

Este modelo estima que la constancia fue de:73.304 de frecuencia cardiaca de 0 años y que una persona que aumenta .0324 de frecuencia cardiaca por cada año que pase

**Prueba de hipotesis para la pendiente**

$$
H0:β1=0
$$
$$
H1:β1≠0
$$

```python
modelo.conf_int(0.05)
```

**TEST DE ANOVA**

```python
modelo.summary()
```

Interpretación esperada:

> Como el valor-p del ANOVA es mayor que 0.05, se acepta ($H_1$), por lo que existe evidencia estadística de que el modelo lineal que si cumple

Ya que nos dio un valor de .068

**Coefcciente de determinacion**

```python
# Coeficiente de determinación
from sklearn.metrics import r2_score

y_calculada = modelo.predict(x_constante)
r2 = r2_score(Y, y_calculada)

print(f"Coeficiente de determinación: {r2: 0.4%}")
```

> El valor de $R^2$ = 0.332 indica que el 33.2% de la variabilidad en la frecuencia cardiaca es explicada por la frecuencia cardiaca utilizando un modelo lineal.

```python
from scipy.stats import shapiro
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import statsmodels.formula.api as smf # Import statsmodels.formula.api

# Fit the linear regression model for Modelo 1: DiastolicBP ~ SystolicBP
modelo = smf.ols('HeartRate ~ Age', data=df).fit()

residuales = modelo.resid
# Prueba de Shapiro-Wilk
stat, valor_p_sh = shapiro(residuales)
print(f"valor-p (Shapiro) = {valor_p_sh}")

# Visualización: Q-Q plot
sm.qqplot(residuales, line='s')
plt.title("Q-Q plot de residuos")
plt.show()

# Histograma
sns.histplot(residuales, kde=True)
plt.title("Histograma de residuos")
plt.show()
```

```python
# Test de Breusch-Pagan
# H0: No hay heterocedasticidad (dispersión es igual en toda la recta)
# H1: Hay Heterocedasticidad (dispersión distinta a lo largo de la recta)
# alpha = 0.05

from statsmodels.stats.api import het_breuschpagan
import statsmodels.api as sm
_, valor_p_bp, _, _ = het_breuschpagan(residuales, x_constante)
print(f'valor-p de Breusch-Pagan: {valor_p_bp: 0.4f}\n')
```

```python
# test de ANOVA (Analysis of Variance)
# H0: beta_1 = 0   (No hay correlación)
# H1: beta_1 ≠ 0   (Sí hay correlación)

import statsmodels.api as sm
from statsmodels.formula.api import ols
# Y ~ X
modelo_lineal = ols(
    '''HeartRate ~Age ''',
    data = df).fit()
tabla_anova = sm.stats.anova_lm(modelo_lineal)
tabla_anova
```
